```python
print("genre unik :", df["genre"].unique())
print("genreL1 unik : ", df["genreL1"].unique())
df["genreL1"].value_counts()
```

```python
df_clean = df.copy()
```

```python
df_clean["created"] = pd.to_datetime(df_clean["created"], errors='coerce')
df_clean["updated"] = pd.to_datetime(df_clean["updated"], errors='coerce')
df_clean["created"] = df_clean["created"].dt.date
```

```python
print("duplikat berdasarkan uid :", df_clean.duplicated(subset=['uid']).sum())
df_clean = df_clean.drop_duplicates(subset=['uid'], keep='first')
print("jumlah baris setelah dihapus :", len(df_clean))
```

```python
df_clean.isnull().sum()
```

```python
df_clean["genreL1"] = df_clean["genreL1"].fillna("Unknown")
```

```python
df_clean["description"] = df_clean["description"].fillna("Unknown")
```

```python
df_clean.isnull().sum()
```

```python
top10 = df_clean.sort_values(by='playing', ascending=False).head(10)
top10[['name','playing','visits','genreL1']]
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.barh(top10['name'], top10['playing'], color = 'skyblue')
plt.xlabel('numbers of players')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
genre_counts = df_clean['genreL1'].value_counts()
print(genre_counts)

plt.figure(figsize=(10,5))
barh = plt.barh(genre_counts.index, genre_counts.values, color='lightgreen')
plt.ylabel('Number of Games')
plt.title('Number of Games by Genre')
plt.tight_layout()
plt.show()
```

```python
total_votes = df_clean['totalUpVotes'] + df_clean['totalDownVotes']
df_clean["rating"] = (df_clean["totalUpVotes"] / total_votes * 100).round(2)
df_clean['rating'] = df_clean['rating'].fillna(0)
df_clean[['name', 'totalUpVotes', 'totalDownVotes', 'rating']].head(10)
```

```python
kolom_angka = ['playing', 'visits', 'likes', 'totalUpVotes', 'totalDownVotes', 'rating']
korelasi = df_clean[kolom_angka].corr()
korelasi
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.imshow(korelasi, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Korelasi")
plt.xticks(range(len(kolom_angka)), kolom_angka, rotation=45, ha="right")
plt.yticks(range(len(kolom_angka)), kolom_angka)
plt.title("Korelasi Antar Metrik Game")
# tulis angka di tiap kotak
for i in range(len(kolom_angka)):
    for j in range(len(kolom_angka)):
        plt.text(j, i, f"{korelasi.iloc[i, j]:.2f}", ha="center", va="center", fontsize=9)
plt.tight_layout()
plt.show()
```

```python
rating_per_genre = df_clean.groupby("genreL1")["rating"].mean().sort_values(ascending=False)
print(rating_per_genre)

plt.figure(figsize=(10, 5))
barh = plt.barh(rating_per_genre.index, rating_per_genre.values, color='orange')
plt.ylabel("Rating Rata-rata (%)")
plt.title("Rating Rata-rata per Genre")
plt.tight_layout()
plt.show()
```
